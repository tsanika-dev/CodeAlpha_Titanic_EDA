## Understanding the Dataset Features

The Titanic dataset contains information about passengers who were aboard the Titanic.

Important features include:

- **PassengerId:** Unique identification number of each passenger.
- **Survived:** Survival status (0 = Did not survive, 1 = Survived).
- **Pclass:** Passenger class (1 = First, 2 = Second, 3 = Third).
- **Name:** Name of the passenger.
- **Sex:** Gender of the passenger.
- **Age:** Age of the passenger.
- **SibSp:** Number of siblings or spouses aboard.
- **Parch:** Number of parents or children aboard.
- **Fare:** Ticket fare paid.
- **Cabin:** Cabin number.
- **Embarked:** Port from which the passenger boarded.

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

## Handling Missing Age Values

The Age column contains missing values. Since age can contain extreme values, the median is used to fill missing values.

The median is preferred because it is less affected by extremely young or old ages.

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df = df.drop("Cabin", axis=1)
```

```python
df.isnull().sum()
```

# Survival count Analysis

```python
df["Survived"].value_counts().plot(kind="bar")

plt.title("Survival Count")
plt.xlabel("Survival Status (0 = Did Not Survive, 1 = Survived)")
plt.ylabel("Number of Passengers")

plt.show()
```

### Observation

The visualization compares the number of passengers who survived with those who did not survive. The analysis shows that the number of passengers who did not survive was greater than the number of passengers who survived.

# Passenger Class Distribution Analysis

```python
df["Pclass"].value_counts().sort_index().plot(kind="bar")

plt.title("Passenger Class Distribution")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()
```

### Observation

The dataset contains passengers from three different classes. Third-class passengers represent the largest group, followed by first-class and second-class passengers.

# Age Distribution of Passengers

```python
df["Age"].plot(kind="hist", bins=20)

plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

### Observation

The age distribution shows that the Titanic passengers belonged to different age groups, with a large number of passengers being young adults.

# Gender Distribution

```python
df["Sex"].value_counts().plot(kind="bar")

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()
```

### Observation

The dataset contains more male passengers than female passengers.

# Fare Distribution

```python
df["Fare"].plot(kind="hist", bins=20)

plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")

plt.show()
```

### Observation

Most passengers paid relatively lower fares, while a smaller number of passengers paid significantly higher fares.

# Survival Rate by Gender

```python
survival_by_gender = df.groupby("Sex")["Survived"].mean()

print(survival_by_gender)
```

```python
survival_by_gender.plot(kind="bar")

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")

plt.show()
```

### Observation

The analysis shows that female passengers had a higher survival rate than male passengers.

# Survival Rate by Passenger Class 

```python
survival_by_class = df.groupby("Pclass")["Survived"].mean()

print(survival_by_class)
```

```python
survival_by_class.plot(kind="bar")

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")

plt.show()
```

### Observation

First-class passengers had a higher survival rate compared with passengers from second and third class.

```python
numeric_df = df.select_dtypes(include=["number"])

correlation_matrix = numeric_df.corr()

correlation_matrix
```

```python
summary = pd.DataFrame({
    "Metric": [
        "Total Passengers",
        "Total Survivors",
        "Overall Survival Rate",
        "Average Age",
        "Average Fare"
    ],
    "Value": [
        len(df),
        df["Survived"].sum(),
        round(df["Survived"].mean() * 100, 2),
        round(df["Age"].mean(), 2),
        round(df["Fare"].mean(), 2)
    ]
})

summary
```

# Correlation Matrix

```python
plt.figure(figsize=(10, 6))

plt.imshow(correlation_matrix, aspect="auto")

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.colorbar()

plt.title("Correlation Matrix")

plt.show()
```

### Observation

The correlation matrix shows the relationships between numerical variables. It helps identify whether changes in one numerical feature are associated with changes in another feature.
